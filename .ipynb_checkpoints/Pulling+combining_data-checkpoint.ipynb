**Pulling Data from Dartfish**

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.support.ui import Select
from selenium.webdriver.chrome.options import Options
import time
from selenium.common.exceptions import ElementClickInterceptedException, TimeoutException, NoSuchElementException, StaleElementReferenceException

# Setup Chrome Driver and options
download_folder = '/Users/lkimball/Desktop/Wesleyan_Tennis_Analytics/Match_csvs'
options = Options()
options.add_experimental_option('prefs', {
    "download.default_directory": download_folder,
    "download.prompt_for_download": False,
    "download.directory_upgrade": True,
    "safebrowsing.enabled": True
})
service = Service(executable_path='/Users/lkimball/Desktop/chromedriver-mac-x64/chromedriver')
driver = webdriver.Chrome(service=service, options=options)

# Login
driver.get('https://www.dartfish.tv/Login')

try:
    username_field = WebDriverWait(driver, 20).until(
        EC.presence_of_element_located((By.ID, 'Email'))
    )
    password_field = driver.find_element(By.ID, 'Password')
    login_button = driver.find_element(By.CSS_SELECTOR, "input[type='submit'][value='Sign in']")

    username_field.send_keys('lucaskimball98@gmail.com')
    password_field.send_keys('LUCASgjoa13!')

    try:
        login_button.click()
    except ElementClickInterceptedException:
        driver.execute_script("arguments[0].click();", login_button)
except TimeoutException as e:
    print("Login element not found within the time limit.")
    # Debugging: print the page source and take a screenshot
    page_source = driver.page_source
    with open('page_source.html', 'w') as file:
        file.write(page_source)
    driver.save_screenshot('login_error_screenshot.png')
    print("Page source saved as page_source.html and screenshot as login_error_screenshot.png")
    driver.quit()
    exit()

time.sleep(5)

# Navigate to matches page and set items per page to 200
matches_url = 'https://www.dartfish.tv/Videos?CR=p98196'
driver.get(matches_url)
WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.ID, "ctl00_MCL_MLister_PaginationControl_ItemsPerPageBottom_ItemCountDD")))
Select(driver.find_element(By.ID, "ctl00_MCL_MLister_PaginationControl_ItemsPerPageBottom_ItemCountDD")).select_by_value('200')
time.sleep(5)

# Function to process each match
def process_match(index):
    retries = 0
    max_retries = 2
    while retries < max_retries:
        try:
            matches = WebDriverWait(driver, 20).until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "a[href^='Player?CR=']")))
            if index < len(matches):
                match = matches[index]
                driver.execute_script("arguments[0].scrollIntoView(true);", match)
                match.click()
                time.sleep(5)  # Allow time for the match page to load

                # Scroll to the bottom of the page to find the download link
                driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
                time.sleep(2)  # Wait for scroll to complete

                # Check for Excel and CSV download links before attempting to download
                try:
                    download_excel = WebDriverWait(driver, 20).until(EC.element_to_be_clickable((By.ID, "ctl00_MCL_CsvPaneLink")))
                    driver.execute_script("arguments[0].scrollIntoView(true);", download_excel)
                    driver.execute_script("arguments[0].click();", download_excel)
                    print(f"Excel download link clicked for match at index {index}.")
                except TimeoutException:
                    print(f"Excel download link not clickable or not found for match at index {index}.")
                    raise Exception("Required element not found")

                try:
                    csv_download_link = WebDriverWait(driver, 20).until(EC.element_to_be_clickable((By.ID, "ctl00_MCL_ExportFileList_ExportMatch")))
                    driver.execute_script("arguments[0].scrollIntoView(true);", csv_download_link)
                    driver.execute_script("arguments[0].click();", csv_download_link)
                    print(f"CSV download link clicked for match at index {index}.")
                except TimeoutException:
                    print(f"CSV download link not clickable or not found for match at index {index}.")
                    raise Exception("Required element not found")

                time.sleep(5)  # Wait for download to complete
                driver.get(matches_url)  # Explicitly navigate back to the matches list page
                time.sleep(5)  # Ensure page loads properly

                # Verify we are back on the matches list page
                WebDriverWait(driver, 20).until(EC.presence_of_element_located((By.ID, "ctl00_MCL_MLister_PaginationControl_ItemsPerPageBottom_ItemCountDD")))
                print(f"Successfully navigated back to matches list page after processing match at index {index}.")
                return True
            else:
                print(f"Match at index {index} not found.")
                return False
        except Exception as e:
            retries += 1
            print(f"Error processing match at index {index}: {str(e)}. Retrying {retries}/{max_retries}...")
            driver.get(matches_url)  # Ensure we are back on the matches list page
            time.sleep(5)  # Ensure page loads properly

            # Verify we are back on the matches list page
            try:
                WebDriverWait(driver, 20).until(EC.presence_of_element_located((By.ID, "ctl00_MCL_MLister_PaginationControl_ItemsPerPageBottom_ItemCountDD")))
            except TimeoutException:
                print(f"Failed to navigate back to matches list page after error processing match at index {index}.")
                driver.get(matches_url)
                time.sleep(5)
    print(f"Skipping match at index {index} after {max_retries} retries.")
    return False

# Process each match individually by its index
match_count = len(WebDriverWait(driver, 20).until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "a[href^='Player?CR=']"))))
for index in range(match_count):
    print(f"Processing match at index {index}...")
    success = process_match(index)
    if not success:
        # Log and continue to the next match if current match fails after retries
        print(f"Skipping match at index {index} due to repeated errors.")
        continue

driver.quit()
```

```python
import pandas as pd
import os

# List all CSV files in the download directory
csv_files = [f for f in os.listdir(download_folder) if f.endswith('.csv') and '(1)' not in f]

# Initialize an empty list to store DataFrames
dfs = []

# Iterate over CSV files, read each into a DataFrame, and append to the list
for file in csv_files:
    df = pd.read_csv(os.path.join(download_folder, file))
    dfs.append(df)

# Combine all DataFrames into a single DataFrame
combined_df = pd.concat(dfs, ignore_index=True)

# Optionally, save the combined DataFrame to a CSV file (for verification or local use)
combined_df.to_csv('/Users/lkimball/Desktop/Wesleyan_Tennis_Analytics/combined_file.csv', index=False)
```
