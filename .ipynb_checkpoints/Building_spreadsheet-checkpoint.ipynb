```python
# To calculate the values for the tennis stats, we first need to understand how the combined_file
# data relates to the tennis stats columns. We'll start with simple aggregations like total points,
# serve points, and return points.


# Since the combined_file data is very detailed, we'll first identify unique player names
# and then calculate basic stats like total points, serve points, and return points.

# Extract unique player names from the combined_file
player_names = combined_file_df['2 - Player Name'].unique()

# Initialize a dictionary to hold calculated stats for each player
player_stats = {name: {} for name in player_names}

# Correcting the total points calculation to include both points won and lost
for player in player_names:
    # Filter data for the current player, either as server or returner
    player_data = combined_file_df[(combined_file_df['2 - Player Name'] == player) |
                                   (combined_file_df['B1: Returner'] == player)]

    # Total points are now the count of rows where the player is either serving or returning
    total_points = len(player_data)
    
    # Serve points are when the player is listed as the server
    serve_points = len(player_data[player_data['A1: Server'] == player])

    # Return points are when the player is listed as the returner
    return_points = len(player_data[player_data['B1: Returner'] == player])

    # Save the calculated stats
    player_stats[player]['total points'] = total_points
    player_stats[player]['serve points'] = serve_points
    player_stats[player]['return points'] = return_points

# Display the corrected stats for the first few players to check
for player, stats in player_stats.items():
    print(f"{player}: {stats}")
    break  # Displaying just the first player's stats for brevity
```

```python
# Creating a new DataFrame to hold the calculated statistics for each player

# Initialize an empty DataFrame with the columns based on 'tennis stats wes'
stats_columns = [
    
  'Name',
  'total points',
  'serve points ',
  'return',
  '%all serves won',
  '% 1st serve points won',
  '1st serves',
  'overall 1st serve in %',
  'deuce 1st serve in %',
  'ad 1st serve in %',
  '1st pts won deuce',
  '1st pts won ad',
  '1st in net',
  '% in net',
  '1st out',
  '% out',
  '1st serve %',
  '1st serve deuce',
  '% deuce',
  '1st serve ad',
  '% ad',
  'aces',
  '1st serve returned in play',
  '% in play',
  'Not returned',
  'error',
  '% not returned',
  '1st returned for winner',
  '% return winner ',
  '1st ret in play deuce',
  'ret err',
  '% in play deuce',
  '1st in play ad',
  'ret errr',
  '% in play ad',
  '% 2nd serve points won',
  '2nd won deuce%',
  '2nd won ad%',
  '2nd serves',
  '2nd in net',
  '%2nd  in net',
  '2nd out',
  '%out.1',
  '2nd serve in ',
  '2nd serve %',
  'Double faults',
  'DF %',
  '2nd serve returned',
  '% in play.1',
  '2nd Not returned',
  '% not returned.1',
  '2nd  returned for winner',
  '% return winner .1',
  'Serve games won',
  'Serve games lost',
  'Serve game win%',
  'Break points saved',
  'Break points lost',
  'Break point save %'
]


new_stats_df = pd.DataFrame(columns=stats_columns)

# Calculate and populate the DataFrame with the stats for each player
for player in player_names:
    player_data = combined_file_df[(combined_file_df['2 - Player Name'] == player) |
                                   (combined_file_df['B1: Returner'] == player)]

    total_points = len(player_data)
    serve_points = len(player_data[player_data['A1: Server'] == player])
    return_points = len(player_data[player_data['B1: Returner'] == player])

    # Create a new row with the calculated stats
    new_row = {
        'Name': player,
        'total points': total_points,
        'serve points': serve_points,
        'return points': return_points
        # Add more stats as calculated
    }

    # Append the new row to the DataFrame
    new_stats_df = new_stats_df.append(new_row, ignore_index=True)

# Display the first few rows of the new DataFrame to verify
new_stats_df.head()
```

```python
import re

# Function to intelligently split player names, considering hyphenated last names and doubles teams
def smart_split_player_names(name, player_data):
    # Check if the name is frequently appearing as a doubles team (more than one hyphen)
    if name.count('-') > 1:
        return name.split('-')
    else:
        # Check if the name is used in doubles context or as a single player
        doubles_context = player_data[(player_data['2 - Player Name'] == name) |
                                      (player_data['B1: Returner'] == name)].shape[0] > 1
        if doubles_context:
            return name.split('-')
        else:
            return [name]

# Reinitialize the DataFrame to correct for doubles and hyphenated names
new_stats_df = pd.DataFrame(columns=stats_columns)

# Process each point in the combined_file
for _, row in combined_file_df.iterrows():
    # This time we use the smart split function to get individual player names
    player_names = smart_split_player_names(row['2 - Player Name'], combined_file_df) + \
                   smart_split_player_names(row['B1: Returner'], combined_file_df)

    for player in player_names:
        # Filter to avoid double counting in case player is both in '2 - Player Name' and 'B1: Returner'
        if player not in row['2 - Player Name'] and player not in row['B1: Returner']:
            continue

        # Update or add the player's statistics
        if player in new_stats_df['Name'].values:
            # Update existing player stats
            player_index = new_stats_df[new_stats_df['Name'] == player].index[0]
            new_stats_df.loc[player_index, 'total points'] += 1
            if player in row['A1: Server']:
                new_stats_df.loc[player_index, 'serve points'] += 1
            else:
                new_stats_df.loc[player_index, 'return points'] += 1
        else:
            # Add new player entry
            serve_points = 1 if player in row['A1: Server'] else 0
            return_points = 1 if player in row['B1: Returner'] else 0
            new_row = {
                'Name': player,
                'total points': 1,
                'serve points': serve_points,
                'return points': return_points
            }
            new_stats_df = pd.concat([new_stats_df, pd.DataFrame([new_row])], ignore_index=True)

new_stats_df.head()
```

```python
# List of all names including doubles and singles
all_names = [
    'Brett Keeling', 'Harrison Collomb', 'Lane Durkin', 'Mohsen-Breen-Durkin',
    'Alexander Bernard', 'Bilski-Mitchell', 'Collomb-Kasday', 'Durkin-Youngberg',
    'Epstein-Soloveychik', 'Legrand-MacDonald', 'Ben Mitchell', 'Mohsen-Breen-Youngberg',
    'Epstein-Youngberg', 'Michael Zheng', 'Alexandra Burnett', 'Leila Epstein',
    'Chase Kasday', 'Lustgarten-MacDonald', 'Patrick Kypson', 'Burnett-Carlos',
    'Carlos-Soloveychik', 'Renna Mohsen-Breen', 'Caitlyn Ferrante', 'Kasday-Collomb',
    'Nicolas Kotzen', 'Bilski-Collomb', 'Collomb-Portnoy', 'Max Lustgarten',
    'Jan Hrazdil', 'Sarah Youngberg', 'JT Bilski', 'Lustgarten-Lee', 'Youngberg-Ferrante',
    'Jackie Soloveychik', 'Carlos Alcaraz Garfia', 'Burnett-Durkin', 'Eudice Chong'
]

# Filter out doubles matches, but ensure "Renna Mohsen-Breen" is included
singles_names = [name for name in all_names if '-' not in name or name == 'Renna Mohsen-Breen']

# Filter the DataFrame to get rows where '2 - Player Name' matches the singles names
singles_matches_df = combined_file_df[combined_file_df['2 - Player Name'].isin(singles_names)]

singles_matches_df  # Display the first few rows of the filtered dataframe
```

```python
# Reinitialize the player stats dictionary to ensure it only includes relevant players
player_stats = {player_name: {'total points': 0, 'serve points': 0, 'return points': 0}
                for player_name in singles_matches_df['2 - Player Name'].unique()}

# Iterate over the dataset to compute serve and return points for each player listed in '2 - Player Name'
for index, row in singles_matches_df.iterrows():
    player_name = row['2 - Player Name']
    server_name = row['A1: Server']
    
    # Update total points for the player
    player_stats[player_name]['total points'] += 1
    
    # Update serve and return points
    if player_name == server_name:
        player_stats[player_name]['serve points'] += 1
    else:
        player_stats[player_name]['return points'] += 1

# Convert the dictionary to a DataFrame for easier viewing and further analysis
refined_stats_df = pd.DataFrame.from_dict(player_stats, orient='index')
refined_stats_df  # Show the first few rows to verify the structure
```

```python
# Refine the calculations using 'F1: Point Won' to determine the winner of the point
for player_name in singles_matches_df['2 - Player Name'].unique():
    player_data = singles_matches_df[singles_matches_df['2 - Player Name'] == player_name]
    serve_points = player_data[player_data['A1: Server'] == player_name]

    # Total points is the count of all rows for the player
    total_points = len(player_data)
    
    # Serve points is the count of rows where the player was serving
    serve_points_count = len(serve_points)

    # first_serves_in and first serve points won
    first_serves_in = len(serve_points[serve_points['A2: Serve Data'].isin(['1st Serve Made', '1st Serve Ace'])])
    first_serve_points_won = len(serve_points[(serve_points['A2: Serve Data'].isin(['1st Serve Made', '1st Serve Ace'])) & (serve_points['F1: Point Won'] == player_name)])
    
    # Aces are now the count of first and second serve aces
    aces = len(serve_points[(serve_points['A2: Serve Data'] == '1st Serve Ace') | (serve_points['A2: Serve Data'] == '2nd Serve Ace')])

    # Update the aces count in the player_stats dictionary
    player_stats[player_name]['aces'] = aces
    
    # Return points is total points minus serve points
    return_points = total_points - serve_points_count

    # Update the player_stats dictionary
    player_stats[player_name] = {
        'total points': total_points,
        'serve points': serve_points_count,
        'return points': return_points,
        '1st serves in': first_serves_in,
        '1st serve points won': first_serve_points_won,
        '% 1st serves won': (first_serve_points_won / first_serves_in) * 100 if first_serves_in > 0 else 0,
        'aces': aces
    }

# Convert the updated dictionary to a DataFrame for easier viewing and further analysis
advanced_stats_df = pd.DataFrame.from_dict(player_stats, orient='index')
advanced_stats_df  # Show the first few rows to verify the structure
```

```python
# Correct the calculation for "% of all serves won"
for player_name in singles_matches_df['2 - Player Name'].unique():
    player_data = singles_matches_df[singles_matches_df['2 - Player Name'] == player_name]
    serve_data = player_data[player_data['A1: Server'] == player_name]
    serve_won_data = serve_data[serve_data['F1: Point Won'] == player_name]

    # Number of serves won
    serves_won = len(serve_won_data)
    total_serves = len(serve_data)

    # Calculate "% of all serves won"
    player_stats[player_name]['% all serves won'] = (serves_won / total_serves) * 100 if total_serves > 0 else 0

# Update the advanced_stats_df DataFrame with the corrected statistics
advanced_stats_df = pd.DataFrame.from_dict(player_stats, orient='index')
advanced_stats_df
```

```python
# Refine the calculation for "1st pts won deuce"
for player_name in singles_matches_df['2 - Player Name'].unique():
    # Filter data for the specific player and when they are serving
    player_data = singles_matches_df[singles_matches_df['2 - Player Name'] == player_name]
    player_serving_data = player_data[(player_data['A1: Server'] == player_name) & 
                                      (player_data['A2: Serve Data'] == '1st Serve Made')]

    # Filter for serves made in the Deuce Court
    deuce_serves_data = player_serving_data[player_serving_data['H2: Deuce Ad'] == 'Deuce Court']

    # Calculate the total and won serves on the Deuce Court
    total_deuce_serves = len(deuce_serves_data)
    won_deuce_serves = len(deuce_serves_data[deuce_serves_data['F1: Point Won'] == player_name])

    # Calculate the percentage of deuce serves won
    player_stats[player_name]['% 1st pts won deuce '] = (won_deuce_serves / total_deuce_serves) * 100 if total_deuce_serves > 0 else 0

# Update the DataFrame with the new calculations
advanced_stats_df = pd.DataFrame.from_dict(player_stats, orient='index')
advanced_stats_df.head()
```

```python
# Add a new statistic for "1st pts won ad" and calculate it
for player_name in singles_matches_df['2 - Player Name'].unique():
    # Filter data for the specific player when they are serving
    player_data = singles_matches_df[singles_matches_df['2 - Player Name'] == player_name]
    player_serving_data = player_data[(player_data['A1: Server'] == player_name) & 
                                      (player_data['A2: Serve Data'] == '1st Serve Made')]

    # Filter for serves made in the Ad Court
    ad_serves_data = player_serving_data[player_serving_data['H2: Deuce Ad'] == 'Ad Court']

    # Calculate the total and won serves on the Ad Court
    total_ad_serves = len(ad_serves_data)
    won_ad_serves = len(ad_serves_data[ad_serves_data['F1: Point Won'] == player_name])

    # Calculate the percentage of Ad serves won
    player_stats[player_name]['1st pts won ad %'] = (won_ad_serves / total_ad_serves) * 100 if total_ad_serves > 0 else 0

# Update the DataFrame with the new calculations
advanced_stats_df = pd.DataFrame.from_dict(player_stats, orient='index')
advanced_stats_df.head()
```

```python
# Calculate the detailed 1st serve fault statistics
for player_name in singles_matches_df['2 - Player Name'].unique():
    player_serving_data = singles_matches_df[(singles_matches_df['2 - Player Name'] == player_name) & 
                                             (singles_matches_df['A1: Server'] == player_name)]

    # Total number of 1st serve faults
    total_faults = player_serving_data[player_serving_data['x: 1st Serve Fault Data'] == 'Fault'].count()[0]

    # Total number of 1st serves
    total_first_serves = player_serving_data[player_serving_data['A2: Serve Data'].str.contains('1st Serve')].count()[0]

    # Fault types
    fault_long = player_serving_data[player_serving_data['x: 1st Serve Fault Placement'] == 'Fault Long'].count()[0]
    fault_net = player_serving_data[player_serving_data['x: 1st Serve Fault Placement'] == 'Fault Net'].count()[0]
    fault_wide = player_serving_data[player_serving_data['x: 1st Serve Fault Placement'] == 'Fault Wide'].count()[0]

    # Calculate percentages
    percent_fault_long = (fault_long / total_first_serves) * 100 if total_first_serves > 0 else 0
    percent_fault_net = (fault_net / total_first_serves) * 100 if total_first_serves > 0 else 0
    percent_fault_wide = (fault_wide / total_first_serves) * 100 if total_first_serves > 0 else 0

    # Update stats dictionary
    player_stats[player_name].update({
        '1st serve faults': total_faults,
        '1st serve attempts': total_first_serves,
        'fault long': fault_long,
        '% fault long': percent_fault_long,
        'fault net': fault_net,
        '% fault net': percent_fault_net,
        'fault wide': fault_wide,
        '% fault wide': percent_fault_wide,
    })

# Update the DataFrame with the new calculations
advanced_stats_df = pd.DataFrame.from_dict(player_stats, orient='index')
advanced_stats_df[['1st serve attempts', '1st serve faults', 'fault long', '% fault long', 'fault net', '% fault net', 'fault wide', '% fault wide']].head()
```

```python
# Calculate serve games won, lost, and win percentage with the updated criteria
for player_name in singles_matches_df['2 - Player Name'].unique():
    player_serving_data = singles_matches_df[(singles_matches_df['2 - Player Name'] == player_name) & 
                                             (singles_matches_df['A1: Server'] == player_name)]

    # Serve games won: 'G2: Game Won' matches '2 - Player Name'
    serve_games_won = player_serving_data[player_serving_data['G2: Game Won'] == player_name].count()[0]

    # Serve games lost: 'G2: Game Won' does not match '2 - Player Name' and is not NaN
    serve_games_lost = player_serving_data[(player_serving_data['G2: Game Won'] != player_name) & 
                                           ~player_serving_data['G2: Game Won'].isna()].count()[0]

    # Total serve games (excluding NaN values)
    total_serve_games = serve_games_won + serve_games_lost

    # Serve game win percentage
    serve_game_win_percentage = (serve_games_won / total_serve_games) * 100 if total_serve_games > 0 else 0

    # Update stats dictionary
    player_stats[player_name]['serve games won'] = serve_games_won
    player_stats[player_name]['serve games lost'] = serve_games_lost
    player_stats[player_name]['serve game win percentage'] = serve_game_win_percentage

# Update the DataFrame with the new calculations
advanced_stats_df = pd.DataFrame.from_dict(player_stats, orient='index')
advanced_stats_df[['serve games won', 'serve games lost', 'serve game win percentage']].head()
```

```python
# Calculate break points saved, lost, and save percentage
for player_name in singles_matches_df['2 - Player Name'].unique():
    # Filter data where the player is serving
    player_serving_data = singles_matches_df[(singles_matches_df['2 - Player Name'] == player_name) & 
                                             (singles_matches_df['A1: Server'] == player_name)]
    
    # Identify break points faced
    break_points_faced = player_serving_data[player_serving_data['zz - Point Type'] == 'Break Point']
    
    # Break points saved (player won the point)
    break_points_saved = break_points_faced[break_points_faced['F1: Point Won'] == player_name].count()[0]
    
    # Break points lost (player lost the point)
    break_points_lost = break_points_faced[break_points_faced['F1: Point Won'] != player_name].count()[0]

    # Calculate break point save percentage
    total_break_points = break_points_saved + break_points_lost
    break_point_save_percentage = (break_points_saved / total_break_points) * 100 if total_break_points > 0 else 0

    # Update stats dictionary
    player_stats[player_name]['break points saved'] = break_points_saved
    player_stats[player_name]['break points lost'] = break_points_lost
    player_stats[player_name]['break point save percentage'] = break_point_save_percentage

# Update the DataFrame with the new calculations
advanced_stats_df = pd.DataFrame.from_dict(player_stats, orient='index')
advanced_stats_df[['break points saved', 'break points lost', 'break point save percentage']].head()
```
